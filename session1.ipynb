```python

num1 = float(input("Enter first number: "))
op = input("Enter operation (+, -, *, /): ")
num2 = float(input("Enter second number: "))

if op == "+":
    result = num1 + num2

elif op == "-":
    result = num1 - num2

elif op == "/":
    if num2 == 0:
        print("Invalid: Cannot divide by zero")
        result = None
    else:
        result = num1 / num2

elif op == "*":
    result = num1 * num2

else:
    print("Invalid operation")
    result = None

if result is not None:
    print("Result =", result)
```

```python

num = int(input("Enter a number: "))

if num <= 1:
    print("Not Prime")
else:
    for i in range(2, num):
        if num % i == 0:
            print("Not Prime")
            break
    else:
        print("Prime")
```
